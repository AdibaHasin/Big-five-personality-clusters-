```python
#classifying +ve and -ve scores in questions

ext_positives = ['EXT1','EXT3','EXT5','EXT7','EXT9']
ext_negatives = ['EXT2', 'EXT4', 'EXT6', 'EXT8', 'EXT10']

est_positives = ['EST1','EST3','EST5','EST6','EST7','EST8','EST9','EST10']
est_negatives = ['EST2','EST4']

agr_positives =['AGR2','AGR4','AGR6','AGR8','AGR9','AGR10']
agr_negatives =['AGR1','AGR3','AGR5','AGR7']

csn_positives = ['CSN1','CSN3','CSN5','CSN7','CSN9','CSN10']
csn_negatives = ['CSN2','CSN4','CSN6','CSN8']

opn_positives = ['OPN1','OPN3','OPN5','OPN6','OPN7','OPN8','OPN9','OPN10']
opn_negatives = ['OPN2','OPN4']

```

```python
def reverse_items(columns, max_scale=5):

    """
    Reverses the Likert scores in the specified columns.

    columns: list of columns to be reversed
    max_scale: Maximum value of the Likert scale
    """

    for col in columns:
        df[col] = (max_scale + 1) - df[col]
```

```python
def scoreCalculator(group,groupname=""):
    df[f"{groupname}_Score"] = df[group].mean(axis=1)
```

```python
reverse_items(ext_negatives)
reverse_items(est_negatives)
reverse_items(agr_negatives)
reverse_items(csn_negatives)
reverse_items(opn_negatives)

scoreCalculator(EXT,"Extroversion")
scoreCalculator(EST,"Neuroticism")
scoreCalculator(AGR,"Agreeableness")
scoreCalculator(CSN,"Conscientiousness")
scoreCalculator(OPN,"Openness")
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# single PCA space for all participants
pca = PCA(n_components=2, random_state=42)
Z = pca.fit_transform(df_model)          # df_model = scaled 5 trait columns (no country)
df_pca = pd.DataFrame(Z, columns=["PCA1","PCA2"])
df_pca["Cluster"] = k_fitted             # KMeans labels
df_pca["country"] = scores_df["country"] # attach country

# limit to top N countries and sample to reduce overplotting
topN = 8
top_countries = df_pca["country"].value_counts().head(topN).index
small = (df_pca[df_pca["country"].isin(top_countries)]
         .groupby("country", group_keys=False)
         .apply(lambda g: g.sample(min(len(g), 3000), random_state=42)))  # ≤3k pts per country

# ompute cluster centroids per country to overlay
centroids = (small.groupby(["country","Cluster"])[["PCA1","PCA2"]]
             .mean().reset_index())

# one panel per country
g = sns.relplot(
    data=small, x="PCA1", y="PCA2", hue="Cluster",
    col="country", col_wrap=4, kind="scatter",
    height=3.5, aspect=1, alpha=0.6, s=10, palette="Set2",
    facet_kws=dict(sharex=True, sharey=True)
)
g.set_titles("{col_name}")
g.set_axis_labels("PCA1", "PCA2")
g._legend.set_title("Clusters")

# Overlay centroids as big markers (✕)
for ax, country in zip(g.axes.flat, g.col_names):
    cc = centroids[centroids["country"] == country]
    ax.scatter(cc["PCA1"], cc["PCA2"], s=160, marker="X", linewidths=1.2, edgecolors="black", alpha=0.9)

plt.suptitle("Personality Clusters after PCA — per Country", y=1.02, fontsize=14)
plt.tight_layout()
plt.show()



```

```python
country_stats = (
    scores_df
    .groupby('country')
    .agg({
        'Extroversion_Score': 'mean',
        'Neuroticism_Score': 'mean',
        'Agreeableness_Score': 'mean',
        'Conscientiousness_Score': 'mean',
        'Openness_Score': 'mean',
        'country': 'count'   # count participants
    })
    .rename(columns={'country': 'Participants'})
    .reset_index()
)
country_stats.head()

```

```python
traits = [
    'Extroversion_Score',
    'Neuroticism_Score',
    'Agreeableness_Score',
    'Conscientiousness_Score',
    'Openness_Score'
]
country_stats = country_stats[country_stats['Participants'] >= 50]
highest = []
for trait in traits:
    idx = country_stats[trait].idxmax()
    highest.append({
        'Trait': trait,
        'Country': country_stats.loc[idx, 'country'],
        'Score': round(country_stats.loc[idx, trait], 3),
        'Participants': country_stats.loc[idx, 'Participants']
    })

summary = pd.DataFrame(highest)
summary
```
